```python
df = pd.read_csv('vehicle_data.csv')
```

```python
data = df.to_dict(orient='records')
```

```python
DB_NAME = "vehicle-insurance"
COLLECTION_NAME = "vehicle-data"
# CONNECTION_URL = "mongodb+srv://vikashdas770:WtwsW3eh6T3J0h6z@cluster0.0aygk.mongodb.net/?retryWrites=true&w=majority&appName=Cluster0"
CONNECTION_URL = os.getenv("MONGO_DB_URL")
```

```python
print(CONNECTION_URL)
print(DB_NAME)
print(COLLECTION_NAME)
```

```python

client = pymongo.MongoClient(CONNECTION_URL)
data_base = client[DB_NAME]
collection = data_base[COLLECTION_NAME]
```

```python
rec = collection.insert_many(data)
```

```python
if rec.inserted_ids:
    print("✅ Data inserted successfully")
    print("Inserted IDs:", rec.inserted_ids)
else:
    print("❌ Insert failed")
```

```python
from pymongo.errors import BulkWriteError

try:
    rec = collection.insert_many(data, ordered=False)
    print(f"✅ Successfully inserted {len(rec.inserted_ids)} documents")

except BulkWriteError as e:
    inserted = e.details.get("nInserted", 0)
    print(f"⚠️ Insert attempted, but only {inserted} documents inserted")

    if inserted == 0:
        print("❌ No data was inserted")
```
